```python
score_cols=["math_score", "history_score", "chemistry_score", "physics_score", "biology_score","english_score","geography_score"]
df["average_score"] = df[score_cols].mean(axis=1)
df["total_score"] = df[score_cols].sum(axis=1)
print(df[["average_score"]].head(30))
```

```python
X=df[["weekly_self_study_hours", "absence_days"]]
Y=df[["average_score"]]
```

```python
X.head()
```

```python
df.shape
```

```python
data=2000
train_size = int(data * 0.8)
test_size = int(data * 0.2)
print("Train size:", train_size)
print("Test size:", test_size)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```
